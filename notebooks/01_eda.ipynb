```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
from src.data_prep import load_data, preprocess_pipeline, prepare_data


```

```python
# Load and explore
df = load_data('../data/raw.csv')
print(df.describe())
print("Missing values:\n", df.isnull().sum())
```

```python
# 1. Univariate distributions
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
numeric_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
for i, col in enumerate(numeric_cols):
    sns.histplot(df[col], kde=True, ax=axes[i//3, i%3])
plt.tight_layout()
plt.savefig('../reports/figures/univariate_dist.png', dpi=300, bbox_inches='tight')
```

```python
# 2. Categorical vs target
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
cat_cols = ['cp', 'restecg', 'slope', 'thal']
for i, col in enumerate(cat_cols):
    crosstab = pd.crosstab(df[col], df['target'], normalize='index')
    crosstab.plot(kind='bar', ax=axes[i//2, i%2])
    axes[i//2, i%2].set_title(f'{col} vs Heart Disease')
plt.tight_layout()
plt.savefig('../reports/figures/cat_target.png', dpi=300)
```

```python
# 3. Correlation heatmap (numeric only)
plt.figure(figsize=(10, 8))
sns.heatmap(df[numeric_cols + ['target']].corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Feature Correlation Matrix')
plt.savefig('../reports/figures/correlation_heatmap.png', dpi=300)
```

```python
# 4. Key insights for report
disease_rate = df['target'].mean()
print(f"Heart disease prevalence: {disease_rate:.1%}")
print("Top correlations with target:")
corr_target = df[numeric_cols + ['target']].corr()['target'].sort_values(ascending=False)
print(corr_target)
```
